```python
"""Assignment 06/12: Implementing Design Patterns I - Creational
Python solutions for all ten questions from the supplied assignment PDF.
Run a section by copying it into a separate .py file, or uncomment/execute
the desired section. Each section has its own demonstration code.
"""

# ============================================================================
# Monday – 1. Singleton: University Configuration System
# ============================================================================
class UniversityConfiguration:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.academic_year = "2026-2027"
            cls._instance.semester = "III/I"
            cls._instance.working_hours = "9:00 AM - 5:00 PM"
            cls._instance.university_name = "SR University"
        return cls._instance

    def display_configuration(self):
        print("University:", self.university_name)
        print("Academic Year:", self.academic_year)
        print("Semester:", self.semester)
        print("Working Hours:", self.working_hours)


config1 = UniversityConfiguration()
config2 = UniversityConfiguration()
config1.display_configuration()
print("Same object:", config1 is config2)

# Sample output:
# University: SR University
# Academic Year: 2026-2027
# Semester: III/I
# Working Hours: 9:00 AM - 5:00 PM
# Same object: True

# Pattern note: Singleton ensures that only one configuration object is created and reused throughout the application.


# ============================================================================
# Monday – 2. Factory Method: Notification System
# ============================================================================
from abc import ABC, abstractmethod

class Notification(ABC):
    @abstractmethod
    def send(self, message):
        pass

class EmailNotification(Notification):
    def send(self, message):
        print("Email notification:", message)

class SMSNotification(Notification):
    def send(self, message):
        print("SMS notification:", message)

class PushNotification(Notification):
    def send(self, message):
        print("Push notification:", message)

class NotificationFactory:
    @staticmethod
    def create_notification(notification_type):
        notification_type = notification_type.lower()
        if notification_type == "email":
            return EmailNotification()
        elif notification_type == "sms":
            return SMSNotification()
        elif notification_type == "push":
            return PushNotification()
        else:
            raise ValueError("Unsupported notification type")

kind = input("Enter notification type (email/sms/push): ").strip()
message = input("Enter message: ")
notification = NotificationFactory.create_notification(kind)
notification.send(message)

# Sample output:
# Enter notification type (email/sms/push): email
# Enter message: Your class starts at 9 AM
# Email notification: Your class starts at 9 AM

# Pattern note: Factory centralizes object creation; the client requests a notification by type instead of constructing a concrete class directly.


# ============================================================================
# Tuesday – 1. Singleton: Library Management System
# ============================================================================
class LibraryManager:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.books = []
            cls._instance.members = []
            cls._instance.transactions = []
        return cls._instance

    def add_book(self, title):
        self.books.append(title)

    def add_member(self, name):
        self.members.append(name)

    def record_transaction(self, details):
        self.transactions.append(details)

    def display(self):
        print("Books:", self.books)
        print("Members:", self.members)
        print("Transactions:", self.transactions)


library1 = LibraryManager()
library1.add_book("Python Programming")
library1.add_member("Ananya")
library1.record_transaction("Ananya borrowed Python Programming")

library2 = LibraryManager()
library2.display()
print("Same manager:", library1 is library2)

# Sample output:
# Books: ['Python Programming']
# Members: ['Ananya']
# Transactions: ['Ananya borrowed Python Programming']
# Same manager: True

# Pattern note: All parts of the application share one LibraryManager and therefore use the same book, member, and transaction lists.


# ============================================================================
# Tuesday – 2. Factory Method: Vehicle Rental System
# ============================================================================
from abc import ABC, abstractmethod

class Vehicle(ABC):
    @abstractmethod
    def rent(self, days):
        pass

class Car(Vehicle):
    def rent(self, days):
        print(f"Car rented for {days} day(s).")

class Bike(Vehicle):
    def rent(self, days):
        print(f"Bike rented for {days} day(s).")

class Bus(Vehicle):
    def rent(self, days):
        print(f"Bus rented for {days} day(s).")

class VehicleFactory:
    @staticmethod
    def create_vehicle(vehicle_type):
        vehicles = {"car": Car, "bike": Bike, "bus": Bus}
        vehicle_class = vehicles.get(vehicle_type.lower())
        if vehicle_class is None:
            raise ValueError("Unsupported vehicle type")
        return vehicle_class()

kind = input("Choose vehicle (car/bike/bus): ").strip()
days = int(input("Number of rental days: "))
vehicle = VehicleFactory.create_vehicle(kind)
vehicle.rent(days)

# Sample output:
# Choose vehicle (car/bike/bus): bike
# Number of rental days: 2
# Bike rented for 2 day(s).

# Pattern note: The factory selects and returns the appropriate Car, Bike, or Bus object based on the customer's choice.


# ============================================================================
# Wednesday – 1. Singleton: Hospital Management System
# ============================================================================
class HospitalManager:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.departments = []
            cls._instance.doctors = []
            cls._instance.services = []
        return cls._instance

    def add_department(self, department):
        self.departments.append(department)

    def add_doctor(self, doctor):
        self.doctors.append(doctor)

    def add_service(self, service):
        self.services.append(service)

    def display(self):
        print("Departments:", self.departments)
        print("Doctors:", self.doctors)
        print("Services:", self.services)


hospital1 = HospitalManager()
hospital1.add_department("Cardiology")
hospital1.add_doctor("Dr. Rao")
hospital1.add_service("Emergency Care")

hospital2 = HospitalManager()
hospital2.display()
print("Same manager:", hospital1 is hospital2)

# Sample output:
# Departments: ['Cardiology']
# Doctors: ['Dr. Rao']
# Services: ['Emergency Care']
# Same manager: True

# Pattern note: The Singleton provides one centralized manager for hospital departments, doctors, and services.


# ============================================================================
# Wednesday – 2. Factory Method: Food Ordering System
# ============================================================================
from abc import ABC, abstractmethod

class FoodOrder(ABC):
    @abstractmethod
    def place_order(self, item):
        pass

class DineInOrder(FoodOrder):
    def place_order(self, item):
        print(f"Dine-in order placed for {item}.")

class TakeawayOrder(FoodOrder):
    def place_order(self, item):
        print(f"Takeaway order placed for {item}.")

class DeliveryOrder(FoodOrder):
    def place_order(self, item):
        print(f"Delivery order placed for {item}.")

class FoodOrderFactory:
    @staticmethod
    def create_order(order_type):
        order_types = {
            "dine-in": DineInOrder,
            "takeaway": TakeawayOrder,
            "delivery": DeliveryOrder
        }
        order_class = order_types.get(order_type.lower())
        if order_class is None:
            raise ValueError("Unsupported order type")
        return order_class()

kind = input("Order type (dine-in/takeaway/delivery): ").strip()
item = input("Food item: ")
order = FoodOrderFactory.create_order(kind)
order.place_order(item)

# Sample output:
# Order type (dine-in/takeaway/delivery): delivery
# Food item: Paneer wrap
# Delivery order placed for Paneer wrap.

# Pattern note: The factory creates the correct order object for dine-in, takeaway, or delivery.


# ============================================================================
# Thursday – 1. Singleton: Online Examination System
# ============================================================================
class ExamManager:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.schedules = []
            cls._instance.subjects = []
            cls._instance.details = []
        return cls._instance

    def add_schedule(self, schedule):
        self.schedules.append(schedule)

    def add_subject(self, subject):
        self.subjects.append(subject)

    def add_exam_detail(self, detail):
        self.details.append(detail)

    def display(self):
        print("Schedules:", self.schedules)
        print("Subjects:", self.subjects)
        print("Exam details:", self.details)


exam1 = ExamManager()
exam1.add_schedule("DSA - 15 October, 10 AM")
exam1.add_subject("Data Structures")
exam1.add_exam_detail("Room 204, 30 marks")

exam2 = ExamManager()
exam2.display()
print("Same manager:", exam1 is exam2)

# Sample output:
# Schedules: ['DSA - 15 October, 10 AM']
# Subjects: ['Data Structures']
# Exam details: ['Room 204, 30 marks']
# Same manager: True

# Pattern note: The Singleton ensures that schedule, subject, and examination information is managed centrally.


# ============================================================================
# Thursday – 2. Factory Method: Document Management System
# ============================================================================
from abc import ABC, abstractmethod

class Document(ABC):
    @abstractmethod
    def open_document(self, filename):
        pass

class PDFDocument(Document):
    def open_document(self, filename):
        print(f"Opening PDF document: {filename}")

class WordDocument(Document):
    def open_document(self, filename):
        print(f"Opening Word document: {filename}")

class ExcelDocument(Document):
    def open_document(self, filename):
        print(f"Opening Excel document: {filename}")

class DocumentFactory:
    @staticmethod
    def create_document(document_type):
        document_types = {
            "pdf": PDFDocument,
            "word": WordDocument,
            "excel": ExcelDocument
        }
        document_class = document_types.get(document_type.lower())
        if document_class is None:
            raise ValueError("Unsupported document type")
        return document_class()

kind = input("Document type (pdf/word/excel): ").strip()
filename = input("Filename: ")
document = DocumentFactory.create_document(kind)
document.open_document(filename)

# Sample output:
# Document type (pdf/word/excel): pdf
# Filename: assignment.pdf
# Opening PDF document: assignment.pdf

# Pattern note: The factory returns the PDF, Word, or Excel document handler according to the selected type.


# ============================================================================
# Friday – 1. Singleton: Banking Management System
# ============================================================================
class BankManager:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
            cls._instance.bank_name = "SR University Bank"
            cls._instance.accounts = {}
            cls._instance.configuration = {"currency": "INR"}
        return cls._instance

    def create_account(self, account_number, balance=0):
        self.accounts[account_number] = balance

    def deposit(self, account_number, amount):
        if amount <= 0:
            print("Amount must be positive.")
        elif account_number not in self.accounts:
            print("Account not found.")
        else:
            self.accounts[account_number] += amount
            print("Deposit successful.")

    def display(self):
        print("Bank:", self.bank_name)
        print("Accounts:", self.accounts)
        print("Configuration:", self.configuration)


bank1 = BankManager()
bank1.create_account("A101", 1000)
bank1.deposit("A101", 500)

bank2 = BankManager()
bank2.display()
print("Same manager:", bank1 is bank2)

# Sample output:
# Deposit successful.
# Bank: SR University Bank
# Accounts: {'A101': 1500}
# Configuration: {'currency': 'INR'}
# Same manager: True

# Pattern note: The Singleton centralizes common bank operations, account data, and configuration in one BankManager.


# ============================================================================
# Friday – 2. Factory Method: Travel Booking System
# ============================================================================
from abc import ABC, abstractmethod

class Booking(ABC):
    @abstractmethod
    def book(self, destination):
        pass

class FlightBooking(Booking):
    def book(self, destination):
        print(f"Flight booked to {destination}.")

class TrainBooking(Booking):
    def book(self, destination):
        print(f"Train booked to {destination}.")

class BusBooking(Booking):
    def book(self, destination):
        print(f"Bus booked to {destination}.")

class BookingFactory:
    @staticmethod
    def create_booking(booking_type):
        booking_types = {
            "flight": FlightBooking,
            "train": TrainBooking,
            "bus": BusBooking
        }
        booking_class = booking_types.get(booking_type.lower())
        if booking_class is None:
            raise ValueError("Unsupported booking type")
        return booking_class()

kind = input("Booking type (flight/train/bus): ").strip()
destination = input("Destination: ")
booking = BookingFactory.create_booking(kind)
booking.book(destination)

# Sample output:
# Booking type (flight/train/bus): train
# Destination: Hyderabad
# Train booked to Hyderabad.

# Pattern note: The factory creates the appropriate FlightBooking, TrainBooking, or BusBooking object.


```
